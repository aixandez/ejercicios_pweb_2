```python
import requests
import pandas as pd
response = requests.get('https://api.escuelajs.co/api/v1/products')
df = pd.DataFrame(response.json())
df
```

```python
#POST para crear un producto nuevo
new_post = {
    'title': 'Buzo rojo universitario',
    'price': 430000,
    'description': 'Buzo rojo universitario creado desde python',
    'categoryId': 2,
    'images': ['https://placeimg.com/640/480/any']
}

response = requests.post('https://api.escuelajs.co/api/v1/products', json=new_post)
print(response.status_code)
print(response.json())
```

```python
#GET para visualizar producto "Buzo rojo universitario"
filtro={'title':'Buzo rojo universitario'}
response = requests.get('https://api.escuelajs.co/api/v1/products', params=filtro)
df = pd.DataFrame(response.json())
df
```

```python
#PUT para cambiar información del producto
datos_actualizados= {
    'title': 'Buzo universitario nuevo',
    'price': 50000,
}
producto_id = 2
response = requests.put(f"https://api.escuelajs.co/api/v1/products/{producto_id}", json=datos_actualizados)
print(response.status_code)
print(response.json())
   
```
